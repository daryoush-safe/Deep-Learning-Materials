# Introduction

## Problem of Overfitting

**Overfitting:**
- Learning the underlying patterns and noise in the training data.
- **High accuracy on training data** but **poor generalization on unseen data**.

**Underfitting:**
- The model is too **simple** to capture the patterns in the data.
- Poor performance on **both training and test data**.

## Bias-Variance Tradeoff

**Motivation:**
- Balancing bias and variance is essential for training neural networks that generalize well to unseen data.
- In neural networks, model complexity can be increased with more layers or neurons, impacting both bias and variance.

**Fundamental Equation:**
$$TrueError = Bias + Variance + IrreducibleNoise$$

**Core Concepts:**
- **Bias:**
    - Systematic error due to overly simplified assumptions.
    - High-bias networks (e.g., shallow networks) may **underfit**, failing to capture complex patterns.
    
- **Variance:**
    - Sensitivity to small changes in training data.
    - High-variance networks (e.g., deep networks) may **overfit**, capturing noise in the data.

**Tradeoff in Neural Networks:**

- **Increasing layers** and parameters can **reduce bias** but **increase variance**.
- **Finding the Balance:** Regularization methods help control this tradeoff by **adjusting model complexity.**

# Regularization

## Core Concepts (L1, L2)

### **Objective**

**Prevent overfitting** by favoring simpler models that avoid fitting noise in training data.

By adding explicit terms to the loss function that inject our inductive biases, we can reduce the gap between training and gap performance.

**Regularized Loss Function:**
$$J(w) = \underbrace{\sum_{n=1}^N loss \left(y^{(n)}, f(y^{(n)}, W)\right)}_{\text{Training Error}} + \underbrace{\lambda R(W)}_{\text{Regularization Term}}$$

- **Training Error (Data Loss):** Model predictions should match training data
- **Regularization:** Prevent the model from doing *too* well on training data

### **L1 (Lasso) Regularization**

**Formula:**
$$\text{L1 Regularization} = \lambda \sum_{j=1}^m |w_j|$$

**Key Properties:**
- **Sparsity:** Forces some weights to exactly zero (feature selection).
- **Use Case:** High-dimensional data with irrelevant features.

### **L2 (Ridge) Regularization**

Adds a penalty for large weights to the loss function
$$\text{L2 Regularization} = \lambda \sum_{j=1}^m \| w_j \|^2 = \lambda W^T W$$

Suitable when all features contribute meaningfully, minimizing overfitting without eliminating features.

$$J(W) = L(W) + \lambda \| W \|^2$$
$$W \leftarrow W - \eta \nabla_W J(W)$$
$$W \leftarrow W - \eta \nabla_W L(W) - 2 \lambda W$$
$$W \leftarrow W \underbrace{(1 - 2\lambda)}_{\text{Weight Decay}} - \eta \nabla_W L(W)$$

*(Where $\text{Wight Decay} \lt 1$ causes gradual weight decay)*

**Use Case:**  
- **Multicollinear features** or when all features are relevant.

**Regularization Parameter ($\lambda$)**

- Controls the strength of the regularization penalty applied to the model's weights.
- **Increasing $\lambda$** places a higher importance on keeping weights small, which:
    - Increases training error (potentially).
    - Reduces overfitting, leading to a simpler and more generalizable model.

### **Elastic Net (L1 + L2)**

Combines both L1 and L2 penalties, where $\beta$ controls the balance between L1 and L2 regularization.
$$\text{ElasticNet Regularization} = \lambda \sum_{j=1}^m (\beta |w_j| + \frac{1 - \beta}{2}w_j^2)$$

Balances feature selection and generalization, effective with high feature correlation.

**Use Case:**
- Correlated features or datasets where both selection and generalization matter.

### **Comparison: L1 vs. L2 Regularization**

| Characteristic             | Ridge Regression                             | Lasso Regression                                       |
|----------------------------|----------------------------------------------|--------------------------------------------------------|
| **Penalty Type**           | L2 (squared coefficient magnitudes)          | L1 (absolute coefficient magnitudes)                   |
| **Coefficient Shrinkage**  | Shrinks coefficients without setting to zero | Can shrink coefficients exactly to zero                |
| **Feature Selection**      | Retains all features                         | Selects features by setting some to zero               |
| **Solution Path**          | Coefficients generally non-zero              | Many coefficients can be zero                          |
| **Model Complexity**       | Includes all features, higher complexity     | Simplifies model by excluding features                 |
| **Impact on Prediction**   | Good for multicollinear data                 | Reduces model complexity, aiding high-dimensional data |
| **Interpretability**       | Less interpretable; all features retained    | More interpretable; excludes irrelevant features       |
| **Best for**               | When all features are relevant               | When identifying key features among many               |
| **Bias-Variance Tradeoff** | Adds bias, reduces variance                  | Similar to Ridge, but with higher bias from exclusion  |
| **Computation**            | Faster without feature selection             | Slightly slower due to feature selection               |

### **Regularization & Activation Functions**

**ReLU:**
- Lower slope for piecewise linear functions
- Avoids saturation (unlike sigmoid/tanh) but can *die* (zero gradients for negative inputs).

**Sigmoid or tanh:** Small $W$ linear regime
- A deep network with small $W$ can also act as a near linear function
- Steep changes facilitate overfitted responses by neurons with large $w$
- **Small weights** $\rightarrow$ Near-linear behavior.
- **Large weights** $\rightarrow$ Risk of steep, overfitted responses.

## Probabilistic Perspective of Regularization

**Probabilistic Modeling of Regression**

* Given observations $D$
* We assign a Gaussian prior to the weights:
    $$p(w) = N(0, \sigma^2)$$
    This means:
    - The weights are assumed to be normally distributed around zero (mean = 0)
    - The variance ($\sigma^2$) controls how strongly we expect the weights to stay near zero.

**Recall: Maximum a Posterior:**
$$\theta_{MAP} = \max_{\theta}p(\theta | D)$$
Since $p(\theta | D) \propto p(D | \theta)p(\theta)$:
$$\theta_{MAP} = \max_{\theta} p(D | \theta)p(\theta)$$
$$\max_{w}p(w|X,y,\sigma^2) \propto p(y|X,w,\sigma^2)p(w) = \max_w \ln p(y|X,w,\sigma^2)$$
Which:
$$p(w) = N(0, \alpha^2I) = (\frac{1}{\sigma\sqrt{2\pi}})^{d+1} e^{-\frac{1}{2\alpha^2}(w^Tw)}$$

$$\max_w \left[\ln p(y|X,w,\sigma^2) = \max_w (\ln p(w|D)) + (\ln p(w))\right] = \max_w \left[-\frac{1}{2\sigma^2}\sum_{i=1}^N(y^{(i)} - w^Tx^{(i)})^2 + -\frac{1}{2\alpha^2} w^Tw\right]$$
$$ = \min_w \left[\frac{1}{\sigma^2}\sum_{i=1}^N(y^{(i)} - w^Tx^{(i)})^2 + \frac{1}{\alpha^2} w^Tw\right]$$

**Equivalent to regularized SSE with $\lambda = \frac{\sigma^2}{\alpha^2}$**

## Implicit Regularization

- SGD generalizes better than gradient descent
    - Smaller batch sizes may perform better than larger ones
    - The inherent randomness allows to reach different parts of the loss function.
- **Implicit regularization:** a preference for some solutions over others
    - Encourages solutions where all the data fits well (so the batch variance is small)
    - Rather than solutions where some of data fit extremely well and other data less well
- SGD implicitly favors places where the **gradients are stable** (where all the batches agree on the slope)

# Early Stopping

**Core Concept**
- **Stop training when validation loss plateaus** to prevent overfitting.
- Prevents overfitting by avoiding excessive training beyond the optimal point.

**Mechanism**
- Acts as an **implicit L2 regularizer**:
    - Weights start small (initialization) and lack time to grow excessively.
    - Limits model complexity by truncating training.

**Implementation**
- Single Hyperparameter **(patience)**: Number of epochs to wait after validation loss stops improving.

**Advantages**
- **Computationally efficient** (no extra terms in loss function).
- **Adaptive** (automatically adjusts to data dynamics).

# Model Ensembles

**Core Concept**  
Combine the predictions from multiple models to create a single, more robust predictor.

**Key Benefits**
- **Improves Generalization:** Creates more robust and reliable predictions on new, unseen data.
- **Reduces Overfitting:** Averages out the unique biases and variances of individual models.
- **Increases Effective Capacity:** The combined "knowledge" of the ensemble can approximate more complex functions than any single model.

**Ensemble Prediction Methods**

- **Regression:**
    - Mean/median of outputs

- **Classification:**
    - Majority vote or mean of pre-softmax activations

**Ensemble Prediction Methods**

- **For Regression Tasks:**
    - Take the **mean** or **median** of all model outputs.

- **For Classification Tasks:**
    - **Majority Voting:** The class with the most votes wins.
    - **Averaging Pre-Softmax Activations:** Often works better than averaging post-softmax probabilities, as it preserves the model's confidence scores more effectively.

**Recall: Bagging (Bootstrap Aggregating)**

A foundational ensemble technique where diversity is created through different data subsets.
- **Bootstrap Sampling:** Create multiple random subsets (with replacement) from the training data.
- **Train in Parallel:** Train a separate model on each of these data subsets.
- **Aggregate Predictions:** For a test instance, all models in the ensemble vote.
- **Result:** Empirically proven to significantly improve performance over a single model trained on all the data.

**Practical Candidates for Ensemble Members**

You can create an ensemble from:

- **Multiple Independent Models:**
    - Train several distinct models from scratch.

- **Same Model, Different Initializations:**
    - After finding the best hyperparameters, train multiple copies using different random seeds. This is simple and often very effective.

- **Top Cross-Validation Models:**
    - Use the best-performing models found during your cross-validation folds.

- **Different Checkpoints of a Single Model:**
    - Use model snapshots from different training epochs. This is a cost-effective method when full re-training is expensive, though it may have limited success.

- **Running Average of Parameters (Polyak Averaging):**
    - Maintain an exponentially decaying average of the network's weights during training. The final "model" is this average, which often performs better than the final weights alone.

# Drop-Out

## What is Dropout?

**Dropout** is a regularization technique used to reduce overfitting in neural networks.
- During training, **randomly deactivates (drops)** a fraction of neurons in each layer.
- Each forward pass effectively trains a **different thinned sub-network**.
- At test time, all neurons are active, and the network behaves as the *average* of these sub-networks.

**Key Effect**

Dropout prevents neurons from co-adapting — forcing them to learn **robust**, **redundant**, and **generalizable** features.

Effect on Representation: Pattern Density:
- Without dropout:
    - A non-compressive layer may simply **copy** its inputs to outputs.
- With dropout:
    - Neurons must encode **richer and more redundant** patterns.
    - Increases **information density** and **representation diversity**.

## Why Dropout Works?

**Pattern Density & Robust Feature Learning**

- Neurons **cannot rely on specific others**, encouraging independent and distributed representations.
- Has a similar effect to $L_2$ regularization, encouraging smaller weights.

**Implicit Model Ensemble**

- Each dropout mask defines a distinct sub-network.
- For a network with $M$ neurons $\rightarrow$ there are $2^M$ possible sub-networks.
- Training with dropout ≈ training an **ensemble** of these sub-models sharing parameters.
- At test time, we approximate the **average prediction** of this ensemble.

## Mathematical Formulation

### **Forward Pass (with dropout)**

Consider:
$$z_i^{(l+1)} = w_i^{(l+1)}{y}^{(l)} + b_i^{(l+1)}$$
$$y_i^{(l+1)} = f(z_i^{(l+1)})$$
Where $f$ is any activation function.

With dropout, the feed-forward operation becomes:
$$
\begin{align*}
r_j^{(l)} &\sim \text{Bernoulli}(p) \\
\widetilde{y}^{(l)} &= m^{(l)} \cdot y^{(l)} \\
z_i^{(l+1)} &= w_i^{(l+1)} \cdot \widetilde{y}^{(l)} + b_i^{(l+1)} \\
y_i^{(l+1)} &= f(z_i^{(l+1)})
\end{align*}
$$
Where:
- $p$: Keep probability

**Implementing Dropout**

> for $l = 1, \cdots, L$:  
> &nbsp;&nbsp;&nbsp;&nbsp;$m^{[l]} \leftarrow I(rand(N, 1) \lt keepProb)$  
> &nbsp;&nbsp;&nbsp;&nbsp;$a^{[l]} \leftarrow a^{[l]} m^{[l]}$  

### **Backward Pass**

#### **Conceptual**

- Gradients are propagated only through active neurons.
- For dropped neurons: gradient = 0.
- The mask $m^{[l]}$ is applied consistently in both forward and backward passes.

#### **Algorithm**

**Output layer ($L$):**
$$-\frac{\partial \text{loss}(a_{i}^{[L]}, y)}{\partial a_{i}^{[L]}}$$
$$-\frac{\partial \text{loss}}{\partial z_{i}^{[L]}} = f_{L}^{\prime} \left( z_{i}^{[L]} \right) \frac{\partial \text{loss}}{\partial a_{i}^{[L]}}$$

**For layer $ l = L - 1 $ downto 0**
- For $ i = 1 ... d^{[l]} $
- if (not dropout layer OR mask(l, i))
  - $\frac{\partial \text{loss}}{\partial a_{i}^{[l]}} = \sum_{j} w_{ij}^{[l+1]} \frac{\partial \text{loss}}{\partial z_{j}^{[l+1]}} mask(l + 1, j)$
  - $\frac{\partial \text{loss}}{\partial z_{i}^{[l]}} = f_{L}^{\prime} \left( z_{i}^{[l]} \right) \frac{\partial \text{loss}}{\partial a_{i}^{[l]}}$
  - $\frac{\partial \text{loss}}{\partial w_{ij}^{[l+1]}} = a_{i}^{[l]} \frac{\partial \text{loss}}{\partial z_{j}^{[l+1]}} mask(l + 1, j)$ for $ j = 1 ... d^{[l+1]}$
- else
  - $\frac{\partial \text{loss}}{\partial z_{i}^{[l]}} = 0$

## Test Phase

### **What Each Neuron Computes?**

**Dropout: Test-Time**

During test-time all neurons are active. Dropout makes our output random! Want to **average out** the randomness at test-time.  
Mathematically:
$$y = f_W(x, z)$$
where:
- $y$: Output (Label)
- $x$: Input (Image)
- $z$: Random mask

Average out random mask:
$$y = f(x) = E_z[f(x, z)] = \int p(z) f(x, z) dz$$

This integral seems hard, we want to approximate it.

Each neuron actually has the following activation:
$$a_{i}^{[l]} = D \phi(\sum_j w_{ji}^{[l]} a_{j}^{[l-1]} + b_{i}^{[l]})$$
Where:
- $D$: Bernoulli variable that takes a value 1 with probability $\alpha$

$D$ may be switched on or off for individual sub networks, but over the ensemble, the expected output of the neuron is:
$$a_{i}^{[l]} = \alpha \phi(\sum_j w_{ji}^{[l]} a_{j}^{[l-1]} + b_{i}^{[l]})$$

During test time, we will use the **expected output** of the neuron
- Which corresponds to the bagged average output
- Consists of simply scaling the output of each neuron by $\alpha$

### **Implementation Approach**

#### **Vanilla Dropout (Not Recommended)**

Scales activations at test time.

```python
""" Vanilla Dropout: Not recommended implementation (see notes below) """

p = 0.5 # probability of keeping a unit active. higher = less dropout

def train_step(X):
    """ X contains the data """
    # forward pass for example 3-layer neural network
    H1 = np.maximum(0, np.dot(W1, X) + b1)
    U1 = np.random.rand(*H1.shape) < p # first dropout mask
    H1 *= U1 # drop!
    H2 = np.maximum(0, np.dot(W2, H1) + b2)
    U2 = np.random.rand(*H2.shape) < p # second dropout mask
    H2 *= U2 # drop!
    out = np.dot(W3, H2) + b3
    
    # backward pass: compute gradients... (not shown)
    # perform parameter update... (not shown)

def predict(X):
    # ensemble forward pass
    H1 = np.maximum(0, np.dot(W1, X) + b1) * p # NOTE: scale the activations
    H2 = np.maximum(0, np.dot(W2, H1) + b2) * p # NOTE: scale the activations
    out = np.dot(W3, H2) + b3
```

#### **Inverted Dropout (Recommended)**

Since test-time performance is critical, it is preferable to use **inverted dropout**
- Performs the **scaling at train** time, leaving the forward pass at test time untouched.
- Just **divide the activation** of each layer by keep-prob

```python
p = 0.5 # probability of keeping a unit active. higher = less dropout

def train_step(X):
    # forward pass for example 3-layer neural network
    H1 = np.maximum(0, np.dot(W1, X) + b1)
    U1 = (np.random.rand(*H1.shape) < p) / p # first dropout mask. Inverted dropout: divide by p
    H1 *= U1 # drop!
    H2 = np.maximum(0, np.dot(W2, H1) + b2)
    U2 = (np.random.rand(*H2.shape) < p) / p # second dropout mask. Inverted dropout: divide by p
    H2 *= U2 # drop!
    out = np.dot(W3, H2) + b3
    
    # backward pass: compute gradients... (not shown)
    # perform parameter update... (not shown)

def predict(X):
    # ensemble forward pass
    H1 = np.maximum(0, np.dot(W1, X) + b1) # no scaling necessary
    H2 = np.maximum(0, np.dot(W2, H1) + b2) # no scaling necessary
    out = np.dot(W3, H2) + b3
```

## Training vs. Inference

**During training:**
- Each neuron is **kept with probability** $p$ (or dropped with probability $1−p$).
- Dropout mask sampled from a **Bernoulli distribution**.
- Gradients flow only through the **on** neurons. for remaining the gradient is just 0.
<div style="text-align:center">
    <img src="../assets/dropout_training.png" alt="dropout during training">
</div>

**During Inference:**
- All neurons are active.
- We approximate the ensemble average by **scaling activations or weights**.
- **Expected Output Approximation**
    - For neuron $i$:
    $$a_{i}^{[l]} = \alpha \phi(\sum_j w_{ji}^{[l]} a_{j}^{[l-1]} + b_{i}^{[l]})$$
    Where $\alpha = p$ (keep probability)

## Practical Notes & Tips

| Concept                   | Recommendation                                                                            |
| ------------------------- | ----------------------------------------------------------------------------------------- |
| **Dropout Rate**          | Start with 0.5 for hidden layers, tune per model                                          |
| **Where to Apply**        | Commonly used in fully connected layers; use **spatial dropout** for convolutional layers |
| **Avoiding Underfitting** | Don’t set dropout too high (causes excessive sparsity)                                    |
| **Different Layers**      | You can use different keep probabilities for different layers                             |
| **Cost Function**         | Dropout changes the effective cost function—no longer strictly well-defined               |
| **Overfitting Control**   | If your NN is overfitting, dropout helps; if not, use a **larger network with dropout**   |
