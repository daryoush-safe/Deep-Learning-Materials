# Activation Functions

## Step Function

**Formula:**
$$
\text{Step}(z) = \begin{cases}
1 & \text{if } z \ge 0 \\ 
0 & \text{if } z \lt 0
\end{cases}
$$

- Not differentiable at $z=0$
- Derivative is $0$ elsewhere $\implies$ **Not suitable**

## Sigmoid

**Formula:**
$$\sigma(z) = \frac{1}{1 + e^{-z}}, \quad \text{Where: } z = \sum_i w_ix_i + b$$

<div style="text-align:center">
    <img src="../assets/sigmoid.png" alt="sigmoid function">
</div>

**Derivative:**
$$\sigma(z)' = \sigma(z) (1 - \sigma(z))$$

<div style="text-align:center">
    <img src="../assets/sigmoid_derivative.png" alt="sigmoid derivative function">
</div>

**Advantages:**
- Squashes the input between 0 and 1, which makes it useful in **probabilistic interpretations** (e.g., logistic regression).
- Often used in output layers for **binary classification** problems.
- **Smooth** & **Differentiable**

**Limitations:**
- **Gradient Saturation:** When $z$ is very large or very small, the gradient becomes nearly zero, causing **slow learning** (**vanishing** gradient problem).
- **Not Zero-Centered:** The output is not zero-centered, which can make optimization more difficult.

### **Problem: Zero-Centered Activation Functions**

**The Core Issue**

- The output range of a sigmoid function is ($0$, $1$).
- This means that for any input, the output of a neuron using sigmoid is **always positive**.

**Why This Matters: The Gradient Problem**

If the input to a neuron ($x$) is **always positive**, it causes an optimization problem during gradient descent.

Let's break down the gradient calculation for a weight $w_i$:
The local gradient $\frac{\partial L}{\partial w} is:
$$
\frac{\partial L}{\partial w} = \sigma(\sum_i w_i x_i + b) \times (1 - \sigma(\sum_i w_i x_i + b)) \times x \times \text{upstream-gradient}
$$

Now, let's analyze the **sign** (positive/negative) of each component:
- $\sigma(\sum_i w_i x_i + b)$: **Always positive** (sigmoid output > 0).
- $(1 - \sigma(\sum_i w_i x_i + b))$: **Always positive** (since sigmoid output < 1).
- $x$: **Always positive** (because it's the output of a previous sigmoid layer).
- upstream-gradient: Can be positive or negative.

**The Consequence: Suboptimal Weight Updates**
- The signs of the first three terms are **always positive**.
- Therefore, the final sign of the gradient for **every weight $w_i$** is determined **solely** by the sign of the upstream-gradient.
- **Result:** The gradients for all weights $w_i$ in a layer will have the **same sign** (all positive or all negative).

This means the weight vector can only be updated in a **zig-zag path** (like moving along a diagonal), rather than the more direct path to the optimum. This makes the optimization process **inefficient and slow**.

**In short: Always-positive outputs lead to constrained, sub-optimal weight updates, slowing down training.**

## Soft-Max

Soft-max vector activation is often used at the output of multi-class classifier networks to convert raw outputs to probabilities for each class:
$$
o_i = \frac{\exp(z_i)}{\sum_j \exp(z_j)}, \quad \text{where } z_i = \sum_j w_{ji} x_j + b_i
$$

Where:
- $o_i$: Probability $P(y = i | x)$

<div style="text-align:center">
    <img src="../assets/soft_max.png" alt="applying soft-max function">
  </div>

**Key Points:**
- Soft-max is used in the output layer for **multi-class classification**
- It converts **logits** (raw unbounded numerical scores) into a **probability distribution** across classes.
- The class with the **highest probability** is **selected** as the prediction.
- It ensures all outputs sum to 1, making it ideal for choosing one class out of multiple options.

## Hyperbolic Tangent (Tanh)

**Formula:**
$$\text{Tanh}(z) = \frac{e^z - e^{-z}}{e^z + e^{-z}}$$

<div style="text-align:center">
    <img src="../assets/tanh_activation.png" alt="Tanh function">
</div>

**Advantages:**
- **Zero-Centered:** Output ranges from -1 to 1, making optimization easier. (Balanced Updates $\rightarrow$ Reduced Bias in Gradient Descent $\rightarrow$ Faster Convergence)
- Better for **hidden layers** than Sigmoid due to **zero-centered** output.

**Limitations:**
- Similar **saturation issues** as Sigmoid: **Large input** values push **gradients towards zero** (**vanishing** gradient problem).

**Tanh vs. Sigmoid**
- The derivative of the Tanh function has a much **steeper slope** at $x = 0$, meaning it provides a **larger gradient** for backpropagation compared to the Sigmoid function.

## ReLU

**Formula:**
$$\text{ReLU}(z) = \max(0, z)$$

<div style="text-align:center">
    <img src="../assets/basic_relu.png" alt="basic ReLU function">
</div>

**Advantages**
- **Faster convergence**: Specially for **deep** network
- Does not saturate positive values, helping to **avoid vanishing** gradient problem
- **Computationally efficient** (Simpler than Sigmoid/Tanh)
- Converges much faster than sigmoid/tanh in practice (e.g. 6x)
- Actually more biologically plausible than sigmoid

**Limitations:**
- Not **zero-centered** output
- **Dead ReLU Problem:** Neurons can become inactive during training, outputting 0 for all inputs if they receive negative values consistently.

## Leaky ReLU

Allows a small, non-zero gradient for negative inputs.
$$\text{LeakyReLU}(z) = \max(\alpha z, z), \quad \alpha = 0.01$$

<div style="text-align:center">
    <img src="../assets/leaky_relu.png" alt="leaky relu function">
</div>

**Advantages:**
- Similar to ReLU
- Does not saturate
- Computationally efficient
- Converges much faster than sigmoid/tanh in practice! (e.g. 6x)
- Helps prevent the **dead ReLU** problem, where neurons stop updating.

## Parametric ReLU (PReLU)

Similar to Leaky ReLU, but the slope for negative inputs $\alpha$ is **learned** during training.
$$\text{PReLU}(z) = \max(\alpha z, z), \quad \text{$\alpha$ is learned}$$

<div style="text-align:center">

</div>

**Advantages:**
- Provides more flexibility by adjusting the slope for negative inputs based on data.

## Exponential Linear Unit (ELU)

**Similar to ReLU for positive** values but **smoother for negative** inputs.

$$
\text{ELU}(z) = \begin{cases}
z & \text{if } z \gt 0 \\ 
\alpha(e^z - 1) & \text{if } z \le 0
\end{cases}, \quad (\text{default } \alpha = 1)
$$

<div style="text-align:center">
    <img src="../assets/elu.png" alt="ELU function">
</div>

**Advantages:**
- All benefits of RELU
- Closer tro **zero-mean** outputs
- Negative saturation regime compared with Leaky ReLU adds some **robustness to noise**
- Provides **faster convergence** and **reduces bias** shift by smoothing negative values.

**Limitations:**
- Computation requires exp()

## Scaled Exponential Linear Unit (SELU)

**Scaled** version of **ELU** 

$$
\text{ELU}(z) = \begin{cases}
\lambda z & \text{if } z \gt 0 \\ 
\lambda \alpha(e^z - 1) & \text{if } z \le 0
\end{cases}
$$

<div style="text-align:center">
    <img src="../assets/selu.png" alt="SELU function">
</div>

**Advantages:**
- Works better for deep networks.
- Self-normalizing property
- Can train deep SELU networks without *BatchNorm*

## Gaussian Error Linear Unit (GELU)

**Core Concept:**  
GELU weights inputs by their probability under a standard Gaussian distribution, creating a smooth, probabilistic version of ReLU.

**Definition:**

$$GeLU(x) = x \cdot \Phi(x)$$
Where:
- $\Phi(x)$: Cumulative distribution function of standard normal distribution
- Each input $x$ is treated as a random variable from $N(0, 1)$

<div style="text-align:center">
    <img src="../assets/gelu.png" alt="GELU function">
</div>


**Mathematical Foundation:**

**Standard Normal Distribution**
**Probability Density Function:**
$$f(x) = \frac{1}{\sigma \sqrt{2 \pi}} e^{-\frac{1}{2}\left(\frac{x - \mu}{\sigma}\right)^2} \xrightarrow[\sigma = 1]{\mu = 0} \frac{1}{\sqrt{2 \pi}} e^{-x^2 / 2}$$

**Cumulative Distribution Function**
$$\Phi(x) = P(X \leq x) = \int_{-\infty}^x \frac{1}{\sqrt{2 \pi}} e^{-t^2 / 2}  dt$$

**Intuition:** Larger values of $x$ have higher probability of being preserved (approaching identity mapping), while smaller values are suppressed.

**Complete GELU Formula**
$$GELU(x) = x \cdot \Phi(x) = x \int_{-\infty}^x \frac{1}{\sqrt{2 \pi}} e^{-t^2 / 2}  dt$$

**Practical Implementation**

- **Using Error Function**

    - The **Gauss Error Function** provides an efficient computation:
    $$erf(x) = \frac{2}{\pi} \int_{0}^{x} e^{-t^2} dt$$
    - **Conversion:**  
    $$\phi(x) = \frac{1 + \text{erf}\left(\frac{x}{\sqrt{2}}\right)}{2}$$
    - **Final expression:**  
    $$GELU(x) = x \cdot \frac{1 + \text{erf}\left(\frac{x}{\sqrt{2}}\right)}{2}$$

- **Efficient Approximations**
    - Since exact computation is computationally expensive, these approximations are used in practice:
    - **High-precision approximation:**
    $$GELU(x) \approx 0.5x \left(1 + \tanh\left[\sqrt{\frac{2}{\pi}}(x + 0.044715x^3)\right]\right)$$
    - **Simpler approximation:**
    $$GELU(x) \approx x \cdot \sigma(1.702x)$$
    *where $\sigma$ is the sigmoid function.*


**Used in BERT, ViT, GPT**

## Soft-Plus

SoftPlus is a **smooth approximation** to the ReLU function and can be used to constrain the output of a machine to always be **positive**.
$$\text{Softplus}(z) = \ln(1 + e^z)$$

<div style="text-align:center">
    <img src="../assets/soft_plus.png" alt="soft-plus function">
</div>

**Advantages:**
- The output is always positive
- The function smoothly increases and is **always differentiable**, **unlike ReLU** (sharp corner at zero)
- For **negative** inputs, the function **approaches zero**, but unlike ReLU, it never exactly reaches zero, **avoiding the problem of dying neurons**

## MaxOut Neuron

For an input vector $x$, Maxout computes the maximum value across multiple linear transformations:
$$\text{Maxout}(x) = \max_{i=1}^k (w_i^T x + b_i)$$
Where $w_i$ and $b_i$ are learned weight vectors and biases for $k$ linear units.

**Intuition**
- Unlike fixed activation functions (e.g., ReLU, sigmoid), Maxout learns the activation function by selecting the maximum output from ( k ) affine transformations.
- Acts as a **piecewise linear** function, approximating any convex function by combining multiple linear segments.
- Generalizes ReLU (when $k=2$, one unit can mimic ReLU by setting one transformation to zero for negative inputs).

**Key Properties**
- **Non-linear:** Despite using linear transformations, the max operation introduces non-linearity.
- **Learnable:** Parameters $w_i$ and $b_i$ are optimized during training, making Maxout adaptive.
- **Dropout Synergy:** Works well with dropout, as shown in the original paper, enhancing regularization and reducing overfitting.
- **Computationally Intensive:** Requires $k$ times more parameters than a single ReLU unit, increasing model complexity.
- **Universal approximator:** With enough pieces, can approximate any continuous function arbitrarily well.

**Advantages:**
- Generalizes ReLU and Leaky ReLU
- Linear Regime! Does not saturate! Does not die!

**Limitations:**
- Doubles the number of parameters/neuron

## Practical Recommendations

- Use **ReLU**. Be careful with your learning rates
- Try out **Leaky ReLU** / **Maxout** / **ELU** / **SELU** / **GELU**
    - To squeeze out some marginal gains
- Use **sigmoid** and **tanh** only for **gating purposes**

# Data Preprocessing: Input Normalization

## The Core Problem

When input features have **very different scales or ranges**, it forces the optimization algorithm to take small, inefficient steps, significantly **slowing down training**.

## The Two-Step Normalization Process

### **Step 1: Remove the Mean (Centering)**

**Goal:** Make the data zero-centered.

**Calculation:**
- Calculate the mean for each input feature $i$:  
    $$\mu_i = -\frac{\sum_n^N x_i^{(n)}}{N}$$
- Subtract it from each data point:  
    $$x_i = x_i - \mu_i$$

**Why this matters (The "Zig-Zag" Problem):**
- If inputs are always positive, the gradients for all weights in a layer will have the **same sign** (all positive or all negative).
- This forces the weight vector to update in a **zig-zag path**, preventing it from taking a more direct route to the optimum.
- **Result:** Constrained, **inefficient, and slow training.**

### **Step 2: Normalize the Variance (Scaling)**

**Goal:** Give all input features the same scale (variance).

**Calculation:**
- Calculate the variance for each feature:  
    $$\sigma^2 = -\frac{\sum_n^N (x_i^{(n)} - \mu_i)^2}{N}$$
- Divide the centered data by the standard deviation:  
    $$x_i = \frac{x_i}{\sigma_i}$$

**Why this matters:**
- Ensures no single feature dominates the learning process due to its scale, leading to more stable and faster convergence.

**Important Caution:** For many modern applications (like images), full PCA or whitening (which decorrelates data) is **not common**. Simpler centering and scaling are preferred.

## Applying to Test Data

- You **must use the exact same $\mu$ and $\sigma^2$ values** that you computed from the **training set** to normalize your test data.
- This prevents information leakage from the test set and ensures the model sees data that has undergone the same transformation.

# Weight Initialization

## Introduction & Motivation

Weight initialization is the process of choosing starting values for the parameters (weights) in a neural network before beginning the iterative optimization process.

**Why Does Initialization Matter?**

During training, weights are updated via backpropagation.
If all weights start **too small**, activations shrink to zero $\rightarrow$ gradients vanish.
If all weights start **too large**, activations explode $\rightarrow$ gradients blow up.
If all weights are **identical**, all neurons learn the same function.

**Importance:**
- Ensures faster convergence during training
- Improves training stability
- Prevents vanishing or exploding gradients
- Enables networks to learn effectively

**Key Question:** How can we initialize weights to maximize learning efficiency and prevent gradient problems?

## The Symmetry Problem

**What Happens with Constant Initialization?**

**Critical Insight:** Neurons with identical connections that are identically initialized will **never diverge**.

If all weights are initialized to the same value (e.g., all zeros or all ones):
- Each neuron in a layer produces **identical outputs**
- All neurons compute the same function
- All neurons receive **identical gradient updates**
- The network cannot learn distinct features

This is called the **symmetry problem**


**Solution:** Initialize weights with small random values to break symmetry, allowing neurons to learn unique features.

## Initialization Methods

### **Zero Initialization**

**Description:**
- All weights are set to zero.

**Code Example:**

```python
W = np.zeros((Din, Dout))
b = np.zeros((Dout,))
```

**Issue:**

* Every neuron in a layer gets the same gradient.
* All neurons perform identical updates $\rightarrow$ network fails to learn distinct features.

**Key Takeaway:**
- **Never use zero initialization** for weights (biases can be zero).

### **Random Initialization (Small Random Numbers)**

**Description:**
- Assign small random values to weights from a probability distribution to **break symmetry**.

**Distribution Options:**
- Normal (Gaussian) distribution →  $ W \sim \mathcal{N}(0, \sigma^2) $
- Uniform distribution →  $ W \sim U(-\varepsilon, \varepsilon) $

**Code Example:**
```python
W = 0.01 * np.random.randn(Din, Dout)
```

**Performance:**
- Works okay for small networks
- Problems with deeper networks

**Problem 1: Weights Too Small (std = 0.01)**

```python
dims = [4096] * 7    # 6-layer network with hidden size 4096
hs = []
x = np.random.randn(16, dims[0])

for Din, Dout in zip(dims[:-1], dims[1:]):
    W = 0.01 * np.random.randn(Din, Dout)
    x = np.tanh(x.dot(W))
    hs.append(x)
```

**What happens?**
- Input ranges are between -1 and 1
- Weights are very small (0.01)
- Pre-activations (x·W) become very close to zero
- Post-activations (after tanh) become near zero
- **Result:** All activations tend to zero for deeper layers
- **Gradients:** dL/dW → 0, **no learning occurs**

**Problem 2: Weights Too Large (std = 0.05)**

```python
dims = [4096] * 7
hs = []
x = np.random.randn(16, dims[0])

for Din, Dout in zip(dims[:-1], dims[1:]):
    W = 0.05 * np.random.randn(Din, Dout)
    x = np.tanh(x.dot(W))
    hs.append(x)
```

**What happens?**
- **Result:** All activations saturate (approach -1 or +1)
- **Gradients:** Local gradients all zero (tanh derivative ≈ 0 at saturation), **no learning**

### **Xavier Initialization (Glorot Initialization)**

**Description:**
- Designed to keep the variance of activations consistent across layers.

**Ideal for:**
- Sigmoid activation function
- Tanh activation function

**Objective:**
- Prevents the shrinking or exploding of signal magnitudes during forward and backward propagation.

**Goal:**
- Keep variance of outputs close to variance of inputs across layers.

**Distributions:**
- Gaussian: $ W \sim \mathcal{N}\left(0, \frac{1}{n_{\text{in}}}\right) $
- Uniform: $ W \sim U\left(-\sqrt{\frac{6}{n_{\text{in}} + n_{\text{out}}}}, \sqrt{\frac{6}{n_{\text{in}} + n_{\text{out}}}}\right) $


**Mathematical Condition**

**Definition and Goal**

Let the output $ y $ be a weighted sum of $ D_{\text{in}} $ inputs:
$$
y = \sum_{i=1}^{D_{\text{in}}} x_i w_i
$$

**Goal:** We want the variance of the output $ y $ to be equal to the variance of a single input $ x_i $, i.e.,
$$
\text{Var}(y) = \text{Var}(x_i)
$$
This helps prevent the gradients from exploding or vanishing during training.

**Assumptions**

To make the derivation tractable, we assume:
*   **Independence:** All $ x_i $ and $ w_i $ are independent and identically distributed (i.i.d.).
*   **Zero Mean:** Both the inputs $ x_i $ and weights $ w_i $ are zero-centered.
    $$
    \mathbb{E}[x_i] = 0 \quad \text{and} \quad \mathbb{E}[w_i] = 0
    $$

**Derivation of the Variance**

We begin by substituting the definition of $ y $ into its variance.

$$
\text{Var}(y) = \text{Var}\left( \sum_{i=1}^{D_{\text{in}}} x_i w_i \right)
$$

Since all $ x_i $ and $ w_i $ are i.i.d., the variance of the sum is the sum of the variances:
$$
\text{Var}(y) = \sum_{i=1}^{D_{\text{in}}} \text{Var}(x_i w_i) = D_{\text{in}} \cdot \text{Var}(x_i w_i)
$$

We now expand $ \text{Var}(x_i w_i) $. For any two random variables, the variance of their product is:
$$
\text{Var}(x_i w_i) = \mathbb{E}[(x_i w_i)^2] - \left(\mathbb{E}[x_i w_i]\right)^2
$$

Given the independence and zero-mean assumptions:
*   $ \mathbb{E}[x_i w_i] = \mathbb{E}[x_i] \mathbb{E}[w_i] = 0 $
*   $ \mathbb{E}[(x_i w_i)^2] = \mathbb{E}[x_i^2 w_i^2] = \mathbb{E}[x_i^2] \mathbb{E}[w_i^2] $

Therefore, the expression simplifies to:
$$
\text{Var}(x_i w_i) = \mathbb{E}[x_i^2] \mathbb{E}[w_i^2]
$$

For a zero-mean variable, $ \mathbb{E}[x^2] = \text{Var}(x) $. Applying this:
$$
\text{Var}(x_i w_i) = \text{Var}(x_i) \cdot \text{Var}(w_i)
$$

Substituting this back into the expression for $ \text{Var}(y) $:
$$
\text{Var}(y) = D_{\text{in}} \cdot \text{Var}(x_i) \cdot \text{Var}(w_i)
$$

**The Final Condition**

Recall our goal is $ \text{Var}(y) = \text{Var}(x_i) $. Substituting into the equation above:
$$
\text{Var}(x_i) = D_{\text{in}} \cdot \text{Var}(x_i) \cdot \text{Var}(w_i)
$$

Assuming $ \text{Var}(x_i) \neq 0 $, we can divide both sides by $ \text{Var}(x_i) $:
$$
1 = D_{\text{in}} \cdot \text{Var}(w_i)
$$

Solving for $ \text{Var}(w_i) $ gives the critical initialization condition:
$$
\boxed{\text{Var}(w_i) = \frac{1}{D_{\text{in}}}}
$$

Gaussian with zero mean and standard deviation: $1 / \sqrt{fan_{in}}$

**Code Example:**
```python
dims = [4096] * 7
hs = []
x = np.random.randn(16, dims[0])

for Din, Dout in zip(dims[:-1], dims[1:]):
    W = np.random.randn(Din, Dout) / np.sqrt(Din)
    x = np.tanh(x.dot(W))
    hs.append(x)
```

### **He Initialization (Kaiming/MSRA Initialization)**

**Description:**
- Designed specifically for neural networks with ReLU activations, accounting for the non-linearity of these functions.

**Ideal for:**
- ReLU and its variants

**Objective:**
- Prevents exponential growth or reduction of input signal magnitudes through layers.

**Why Different from Xavier?**
- ReLU zeros out half of the neurons (negative values → 0), so the so the effective variance is halved.

**Mathematical Condition**
$$
Var[w] = \frac{2}{n_l}
$$

**Distributions:**
- Gaussian: $ W \sim \mathcal{N}(0, \frac{2}{n_{\text{in}}}) $
- Uniform: $ W \sim U\left(-\sqrt{\frac{6}{n_{\text{in}}}}, \sqrt{\frac{6}{n_{\text{in}}}}\right) $

**Why the Factor of 2?**

**Reasoning:**
- ReLU activation: $f(x) = max(0, x)$
- For a zero-centered input distribution, ReLU zeros out approximately 50% of values
- The variance of the output is reduced by half
- To compensate, we need $\text{Var}(w) = \frac{2}{D_{in}}$ instead of $\frac{1}{D_{in}}$

**Mathematical justification:**
For ReLU (zero-centered $x$):
$$E[ReLU(x)] \approx \frac{E[x]}{2}$$
$$\text{Var}[\text{ReLU}(x)] \approx \frac{\text{Var}(x)}{2}$$

To maintain $\text{Var}(y) = \text{Var}(x)$:
$$D_in · \text{Var}(w) · (\frac{\text{Var}(x)}{2}) = \text{Var}(x)$$
Therefore:
$$\text{Var}(w) = \frac{2}{D_{in}}$$

Gaussian with zero mean and standard deviation: $2 / \sqrt{fan_{in}}$

**Code Example:**
```python
dims = [4096] * 7
hs = []
x = np.random.randn(16, dims[0])

for Din, Dout in zip(dims[:-1], dims[1:]):
    W = np.random.randn(Din, Dout) * np.sqrt(2.0 / Din)
    x = np.maximum(0, x.dot(W))  # ReLU activation
    hs.append(x)
```

### **Summary**

**Decision Guide: Which Initialization to Use?**

**Scenario 1: Deep Network with ReLU**
**Best Choice:** He Initialization
- **Code:** `W = np.random.randn(Din, Dout) * np.sqrt(2.0 / Din)`
- **Reason:** Maintains gradient flow through layers by accounting for ReLU's zeroing behavior

**Scenario 2: Network with Sigmoid/Tanh**
**Best Choice:** Xavier Initialization
- **Code:** `W = np.random.randn(Din, Dout) / np.sqrt(Din)`
- **Reason:** Keeps variance balanced, crucial for symmetric activation functions

**Summary Table**

| Method | Formula | Standard Deviation | Best For | Key Feature |
|--------|---------|-------------------|----------|-------------|
| **Zero** | W = 0 | 0 | ❌ Never | Symmetry problem |
| **Small Random** | W ~ N(0, 0.01²) | 0.01 | Small networks only | Simple but limited |
| **Xavier** | W ~ N(0, 1/n) | √(1/fan_in) | Tanh, Sigmoid | Maintains variance |
| **He** | W ~ N(0, 2/n) | √(2/fan_in) | ReLU, LeakyReLU | Accounts for ReLU zeroing |
